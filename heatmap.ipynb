```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('product_data.csv')
df
```

```python
# In simple the Heatmap is shown the correlation between numerical columns 
# First i have to collect the numerical columns from the dataset
```

```python
numerical = df[['price','rating','stock','discountPercentage','weight']]
print(numerical)
```

```python
# compute correlation 
corr = numerical.corr()
print(corr)
```

```python
#demonstrate with plot
plt.figure(figsize=(12,7))
sns.heatmap(
    corr,
    annot = True ,# show values of correlation
    cmap ='coolwarm',# theme/colour
    fmt ='.2f', # show 2 decimal places
    linewidths=0.5)
plt.title("Heat Map")
```

```python
# Information about heat map values
#What do the numbers mean?
#1.00 → Perfect positive relationship (a column with itself).
#0.80 → Strong positive relationship.
#0.50 → Moderate positive relationship.
#0.00 → No relationship.
#-0.50 → Moderate negative relationship.
#-1.00 → Perfect negative relationship
```
